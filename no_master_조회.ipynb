```python
from src.databases import OracleCRUD, PostgresCRUD
```

```python
puma = PostgresCRUD.select_query(
    """
    select dri.tech_prod, dri.tech, dri.prod_group, dri.level, dri.layer, dri.mode,
	dri.recipe, dri.equip, dri.master_key, drp.para_nm, drp.para_value
    from f1.dm_recipe_info dri
    left join f1.dm_recipe_para drp
    on (dri.recipe_key = drp.recipe_key)
    where dri.model=%s;
    """,
    ("PUMA",),
)
```

```python
kla2365 = PostgresCRUD.select_query(
    """
    select dri.tech_prod, dri.tech, dri.prod_group, dri.level, dri.layer, dri.mode,
	dri.recipe, dri.equip, dri.master_key, drp.para_nm, drp.para_value
    from f1.dm_recipe_info dri
    left join f1.dm_recipe_para drp
    on (dri.recipe_key = drp.recipe_key)
    where dri.model=%s;
    """,
    ("KLA2365",),
)
```
