```python
import os

def remove_empty_folders(target_path):
    # 폴더 존재 여부 확인
    if not os.path.exists(target_path):
        print(f"오류: '{target_path}' 경로가 존재하지 않습니다.")
        return

    count = 0
    # topdown=False 설정으로 가장 깊은 하위 폴더부터 역순으로 탐색합니다.
    # 이렇게 해야 '빈 폴더를 품고 있던 상위 폴더'까지 차례대로 지울 수 있습니다.
    for root, dirs, files in os.walk(target_path, topdown=False):
        # 지정한 최상위 타겟 폴더 자체는 삭제되지 않도록 방지
        if root == target_path:
            continue
            
        # 현재 폴더에 파일과 하위 폴더가 모두 없다면 빈 폴더입니다.
        if not os.listdir(root):
            try:
                os.rmdir(root)
                print(f"삭제됨: {root}")
                count += 1
            except OSError as e:
                print(f"실패: {root} (이유: {e})")

    print(f"\n작업 완료! 총 {count}개의 빈 폴더를 삭제했습니다.")

# ⚠️ 사용 예시 (본인의 폴더 경로로 변경하세요)
# 윈도우 환경 예시: r"C:\Users\Username\Documents\TargetFolder"
# 맥/리눅스 환경 예시: "/Users/Username/Documents/TargetFolder"
target_folder_path = r"D:\DM_DATA_2\SMDI02" 

remove_empty_folders(target_folder_path)
```
